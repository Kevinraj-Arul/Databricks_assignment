```python
# Paths (simulate DBFS)
employee_path = "/Volumes/assignment/datasets/data/Employee-Q1.csv"
department_path = "/Volumes/assignment/datasets/data/Department-Q1.csv"
country_path = "/Volumes/assignment/datasets/data/Country-Q1.csv"

# Read CSVs
employee_df = spark.read.csv(
    employee_path,
    header=True,
    inferSchema=True
)
department_df = spark.read.csv(
    department_path,
    header=True,
    inferSchema=True
)
country_df = spark.read.csv(
    country_path,
    header=True,
    inferSchema=True
)

# Write to bronze layer
# Write to bronze layer with overwrite mode
employee_df.write.csv(
    "/Volumes/assignment/default/writepath/employee",
    mode="overwrite"
)
department_df.write.csv(
    "/Volumes/assignment/default/writepath/department/",
    mode="overwrite"
)
country_df.write.csv(
    "/Volumes/assignment/default/writepath/country/",
    mode="overwrite"
)

display(employee_df)
display(department_df)
display(country_df)
```
